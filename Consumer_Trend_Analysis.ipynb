```python
plt.figure(figsize=(10,6),dpi=200)
plt.subplot(2,2,1)
sns.histplot(df[df['Category']=='Clothing']['Review Rating'],bins=20,kde=True)
plt.subplot(2,2,2)
sns.histplot(df[df['Category']=='Footwear']['Review Rating'],bins=20,kde=True)
plt.subplot(2,2,3)
sns.histplot(df[df['Category']=='Outerwear']['Review Rating'],bins=20,kde=True)
plt.subplot(2,2,4)
sns.histplot(df[df['Category']=='Accessories']['Review Rating'],bins=20,kde=True)
plt.show()
```

```python
# So from the above distribution plots we can see that the distribution of review artings is almost symmetric in all categories.So we can fill the null values in ratings columns with respective medians of the category
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isna().sum()
```

```python
# Renaming columns according to snake casing for better readability and documentation

df.columns = df.columns.str.lower().str.replace(' ','_')
```

```python
df.columns = df.columns.str.strip()
df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'}, inplace=True)
```

```python
df.columns
```

```python
bins = [18,30,45,60,70]
labels = ['Young','Adult','Middle_Age','Senior']
df['age_category'] = pd.cut(df['age'],bins=bins,labels=labels,include_lowest=True)
df[['age','age_category']]
```

```python
# create new column purchase_frequency_days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
df
```
